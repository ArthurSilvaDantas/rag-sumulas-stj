# Coleta das fontes

Este notebook baixa as duas fontes oficiais para `data/raw/` e não altera nada nelas. As súmulas vêm de um PDF e os precedentes qualificados, de um CSV.

| Fonte | Conteúdo | O que mais traz |
|---|---|---|
| `VerbetesSTJ.pdf` (SCON) | Enunciados das súmulas | Marcas de cancelada, revogada e alterada |
| `temas.csv` (Dados Abertos, CC-BY) | Temas, Controvérsias, IAC, PUIL e SIRDR | Tese, questão e situação |

Use a URL do PDF que está abaixo. Existe uma cópia antiga em `www.stj.jus.br/docs_internet/VerbetesSTJ.pdf` que para na Súmula 673 e, por isso, não tem a Súmula 676, que trata da prisão preventiva de ofício.

```python
from pathlib import Path

import requests

ROOT = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / "pyproject.toml").exists())
RAW = ROOT / "data/raw"
RAW.mkdir(parents=True, exist_ok=True)

URL_SUMULAS = "https://scon.stj.jus.br/docs_internet/jurisprudencia/tematica/download/SU/Verbetes/VerbetesSTJ.pdf"
URL_TEMAS = (
    "https://dadosabertos.web.stj.jus.br/dataset/4238da2f-c07b-4c1a-b345-4402accacdcf"
    "/resource/df29da13-7d6b-41ba-ad96-cd1a5bbd191c/download/temas.csv"
)
ATUALIZAR = False  # True baixa as fontes de novo


def baixar(url: str, destino: Path) -> None:
    if destino.exists() and not ATUALIZAR:
        print(f"{destino.name}: usando cópia local")
        return
    resposta = requests.get(url, timeout=120)
    resposta.raise_for_status()
    destino.write_bytes(resposta.content)
    print(f"{destino.name}: {len(resposta.content):,} bytes, Last-Modified {resposta.headers.get('last-modified')}")


baixar(URL_SUMULAS, RAW / "VerbetesSTJ.pdf")
baixar(URL_TEMAS, RAW / "temas.csv")
```
